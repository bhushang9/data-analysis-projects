## Data Visualization

```python
#histogram of numeric columns
df.hist(figsize=(12, 8), edgecolor='black')
plt.suptitle("Histogram of Numerical Features", fontsize=16)
plt.show()
```

```python
#countplots of categorical columns
#Model Distribution 
plt.figure(figsize=(10, 6))
sns.countplot(y=df['model'], order=df['model'].value_counts().index)
plt.title("Model Distribution")
plt.show()
```

```python
#Transmission Distribution
sns.countplot(x="transmission", data=df)
plt.title("Transmission Types")
plt.show()
```

```python
#fuel Type Distribution pie chart
df['fuelType'].value_counts().plot.pie(autopct="%1.1f%%", figsize=(6,6))
plt.title("Fuel Type Distribution")
plt.ylabel("")
plt.show()
```

```python
#scatter plot of Price vs Engine Size
sns.scatterplot(x="engineSize", y="price", data=df, hue="fuelType")
plt.title("Price vs Engine Size")
plt.show()
```

```python
#boxplot of Year vs Price
sns.boxplot(x="year", y="price", data=df)
plt.xticks(rotation=45)
plt.title("Price Distribution Across Years")
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
